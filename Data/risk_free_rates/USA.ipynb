Для США данные берутся из FED (https://www.federalreserve.gov/data/nominal-yield-curve.htm), затем для вычисления риск фри ставки на нетипичные даты используется формула Нельсона-Сигеля (документацию можно найти тут https://www.federalreserve.gov/pubs/feds/2006/200628/200628pap.pdf)

```python
import requests
```

```python
url = "https://www.federalreserve.gov/data/yield-curve-tables/feds200628.csv"

r = requests.get(url)
r.raise_for_status()

with open("risk_free_USA.csv", "wb") as f:
    f.write(r.content)
```
