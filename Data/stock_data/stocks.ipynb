```python
import pandas as pd
import yfinance as yf
```

# Yahoo

```python
def download_stocks(tickers, start):
    for ticker in tickers:
        df = yf.Ticker(ticker).history(start=start, interval="1d", auto_adjust=False)
        df = df[["Open", "High", "Low", "Close", "Adj Close", "Volume", "Dividends"]]
        df.index = df.index.tz_localize(None)
        df.to_csv(f"{ticker}.csv")
```

```python
download_stocks(
    tickers=["BA"],
    start="2016-01-01"
)
```

```python
download_stocks(
    tickers=["^SP500TR"],
    start="2016-01-01"
)
```
